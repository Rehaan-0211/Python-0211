```python
a = np.arange(1,31)
a
```

```python
# a.reshape(5,6) # we have to specify the rows and columns acc to the no of elements we have. This is reshaping
a = a.reshape(5,6) 
```

```python
# Create an array with 23 elements
a = np.arange(23)
print("Original array (23 elements):")
print(a)

# Method 1: Using np.resize() - repeats elements to fill the shape
a_resized = np.resize(a, (5, 6))
print("\nResized to (5,6) using np.resize():")
print(a_resized)

# Method 2: Using array.resize() - fills with zeros (modifies in-place)
a2 = np.arange(23)
a2.resize((5, 6))
print("\nResized to (5,6) using array.resize():")
print(a2)

#Method 1 calls np and asks it to resize the array and method 2 asks the array to fill itself
```
