```python
# **NUMPY INDEXING AND SLICING OF VECTORS**
```

```python
import numpy as np
```

```python
a = np.arange(11,21)
a
```

```python
# These values will have an index starting from zero and we can access them with their index like below:-
a[3]
```

```python
#Slicing
a[1:5:1] # This is called slicing we can manipulate the steps and low high values respectively
a[::] # If we provide no limit then the whole array gets sliced out
```

```python
# NUMPY INDEXING AND SLICING OF A MATRIX
```

```python
ar = np.arange(1,31).reshape(6,5)
ar
```

```python
ar[1] # slices rows
ar[:,1] # slices columns, here : means all rows
```

```python
ar[3:,3:] #slicing a portion of the matrix
```

```python
# BOOLEAN INDEXING
```

```python
arr = np.arange(11,21)
arr
```

```python
bool_index = arr % 2 == 0
bool_index # basically we didnt need to run a loop here for checking if the value is even or %notebook
# we asked which values satisfied the condition and then printed them in the next cell
```

```python
arr = arr[bool_index]
arr
```
