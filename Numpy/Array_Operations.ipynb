```python
#Deep and Shallow copy
```

```python
a = np.arange(1,21)
a
```

```python
slice = a[:5]
slice
```

```python
a[:5] * 10
```

```python
#These are all shallow copies the value of a itself does not change when we apply such operations
```

```python
b = a
#This here is a deep copy.
b[0] = 10
```

```python
a
# Changes the first element for a as well
```

```python
c= a.copy() # This creates a shallow copy thou
c
```

```python
c[0] = 1
c
```

```python
a
#Value does not change for a...since c was a shallow copy
```

```python
# Stacking Array
```

```python
# You can only stack vectors but not matrix 
```

```python
a = np.array([1,3,4,5])
b = np.array([5,6,4,7])

np.vstack((a,b)) # stacks the array row-wise
```

```python
np.hstack((a,b)) # stacks both arrays as one 1D array
```

```python
np.column_stack((a,b)) 
# stacks two arrays column-wise
```

```python
#Splitting Array

g = np.arange(16).reshape(4,4)
g
```

```python
np.hsplit(g,2) # we have to perform equal splits cause if equal split request is compromised we will get an error
```

```python
np.vsplit(g,4)
```

```python
a= np.vsplit(g,2)
```

```python
for i in a:
    print(i) # we can store these split values in a variable to access it separately as well but yea the array remains 2D either way
```
